```python
import random

bot = random.randint(1, 100)

while True:

    player = int(input("Введите число от 1 до 100: "))


    if player < 1 and player > 100:
        print("Число должно быть в диапазоне от 1 до 100.")\
        break

    elif player > bot:
        print("Загаданное число меньше")

    elif player < bot:
        print("Загаданное число больше")

    else:
        print("Угадал")
        break
```
